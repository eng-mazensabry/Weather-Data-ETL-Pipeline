```python
import requests
```

```python
list_of_cities = []
```

```python
cities = [
    "Cairo",
    "Giza",
    "Alexandria",
    "Port Said",
    "Suez",
    "Ismailia",
    "Mansoura",
    "Tanta",
    "Luxor",
    "Aswan"
]
```

```python
url = "https://api.openweathermap.org/data/2.5/weather"

params = {
    "appid": "6a13e61e853763292101962a026aa3e8",
    "q": "Giza"
    
}

response = requests.get(url, params=params)

data = response.json()
data
```

```python
for city in cities:
    print(city)
    params = {
    "appid": "6a13e61e853763292101962a026aa3e8",
    "q": city
    }
    response = requests.get(url, params=params)
    data = response.json()
    city = city
    country = 'Egypt'
    lat = data['coord']['lat']
    lon = data['coord']['lon']
    temp = data['main']['temp']
    pressure = data['main']['pressure']
    Humidity = data['main']['humidity']
    wind_speed = data['wind']['speed']
 
 
    list_of_cities.append({
        'city': city,
        'country': country,
        'lat': lat,
        'lon': lon,
        'temp': temp,
        'pressure': pressure,
        'humidity': Humidity,
        'wind_speed': wind_speed
    })
```

```python
list_of_cities
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(list_of_cities)
df 
```

```python
def fun1(city):
    if city == 'Cairo':
        return 'Greater Cairo'
    elif city == 'Giza':
        return 'Greater Cairo'
    elif city == 'Alexandria':
        return 'North Coast'
    elif city == 'Port Said':
        return 'North Coast'
    elif city == 'Suez':
        return 'Canal'
    elif city == 'Ismailia':
        return 'Canal'
    elif city == 'Mansoura':
        return 'Delta'
    elif city == 'Tanta':
        return 'Delta'
    elif city == 'Luxor':
        return 'Lower Egypt'
    elif city == 'Aswan':
        return 'Lower Egypt'
```

```python
df['Region']= df['city'].apply(fun1)
df
```

```python
df['temp_celsius'] = df['temp'].apply(lambda temp: temp - 273.0)
df
```

```python
def fun2 (temp_celsius):
    if temp_celsius < 30 :
       return 'Cold'
    elif temp_celsius < 34 :
       return 'Moderate'
    else :
       return 'Hot'
```

```python
df['TemperatureCategory'] = df['temp_celsius'].apply(fun2)
df
```

```python
def fun3 (humidity):
    if humidity < 30 :
       return 'Low'
    elif humidity < 40 :
       return 'Medium'
    else :
       return 'High'
```

```python
df['HumidityCategory'] = df['humidity'].apply(fun3)
df
```
